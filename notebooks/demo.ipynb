## 1. 环境准备


```python
# 导入必要的库
import sys
import os
import asyncio

# 添加项目根目录到路径
sys.path.insert(0, os.path.abspath('..'))

from backend.agents.sci_agent_system import get_sci_agent_system
from IPython.display import Markdown, display

# 创建科研智能体系统实例
system = get_sci_agent_system()

print("✅ Universal-SciAgent 1.0 初始化完成")
print("📦 VeADK 原生框架已加载")
print("🔗 火山方舟 API 已连接")
print("")
print("VeADK 组件:")
print("  - Agent + Runner: 智能体核心")
print("  - SequentialAgent: 顺序编排")
print("  - ParallelAgent: 并行编排")
print("  - ShortTermMemory: 短期记忆")
print("  - LongTermMemory: 长期记忆")
```

## 2. 文献综述示例

输入一个研究主题，系统将自动生成文献综述报告。


```python
# 文献综述 - 修改 topic 为您感兴趣的研究主题
topic = "大语言模型在科研自动化中的应用"
domain = "computer_science"  # 可选: biomedical, materials_science, physics, chemistry

print(f"📚 研究主题: {topic}")
print(f"🔬 研究领域: {domain}")
print("\n⏳ 正在生成文献综述，请稍候...")

# 运行文献综述
result = await system.literature_review(
    topic=topic,
    domain=domain,
    max_papers=10
)

print("\n✅ 文献综述完成！")
display(Markdown(str(result.output)))
```

## 3. 假设生成示例

基于文献背景，生成创新性研究假设。


```python
# 假设生成 - 输入文献背景
literature_context = """
当前大语言模型在科研领域的应用面临以下挑战：
1. 领域知识的准确性和可靠性
2. 多模态数据（文本、图像、表格）的理解
3. 长文档的处理能力
4. 科学推理的可解释性

现有研究表明：
- RAG 技术可以增强模型的知识准确性
- 多模态模型在图表理解方面有进展
- 思维链 (CoT) 可以提升推理能力
"""

print("💡 正在生成研究假设...")

result = await system.generate_hypothesis(
    literature_context=literature_context,
    domain="computer_science"
)

print("\n✅ 假设生成完成！")
display(Markdown(str(result.output)))
```

## 4. 实验设计示例

为研究假设设计详细的验证实验方案。


```python
# 实验设计 - 输入要验证的假设
hypothesis = """
假设：结合 RAG 技术与思维链 (CoT) 的多模态科研助手，
在科学文献理解和假设生成任务上的准确率将比单独使用任一技术提高 15-20%。

预期：
- 在 ScienceQA 基准测试上准确率达到 85% 以上
- 在科学论文摘要生成任务上 ROUGE-L 分数达到 0.45 以上
"""

print("🧪 正在设计实验方案...")

result = await system.design_experiment(
    hypothesis=hypothesis,
    domain="computer_science"
)

print("\n✅ 实验设计完成！")
display(Markdown(str(result.output)))
```
